```python
import math

# 상수 정의
g = 9.8  # 중력가속도

# 초기값 설정
v0 = 100.0           # 초기 속도 (m/s)
angle_deg = 37.0     # 발사 각도 (도)
y0 = 140.0           # 초기 높이 (m)

# 각도 라디안으로 변환
angle_rad = math.radians(angle_deg)

# 초기 속도의 x, y 성분
v0x = v0 * math.cos(angle_rad)
v0y = v0 * math.sin(angle_rad)

# 1. 착지 시간 구하기 (이차방정식 풀이)
a = -0.5 * g
b = v0y
c = y0

discriminant = b**2 - 4 * a * c
if discriminant < 0:
    print("해가 없습니다.")
    exit()

t1 = (-b + math.sqrt(discriminant)) / (2 * a)
t2 = (-b - math.sqrt(discriminant)) / (2 * a)
t = t1 if t1 > 0 else t2

print(f"(a) 착지 시간: {t:.2f} 초")

# 2. 수평 거리
x = v0x * t
print(f"(b) 수평 거리: {x:.2f} m")

# 3. 충돌 시 속도 성분
vx = v0x
vy = v0y - g * t
print(f"(c) 충돌 시 속도 성분:")
print(f"    vx: {vx:.2f} m/s")
print(f"    vy: {vy:.2f} m/s")

# 속도 크기
v = math.sqrt(vx**2 + vy**2)
print(f"    속도 크기: {v:.2f} m/s")

# 4. 수평선과 이루는 각도
theta = math.degrees(math.atan2(vy, vx))
print(f"(d) 수평선과 이루는 각도: {theta:.2f} 도")
```
